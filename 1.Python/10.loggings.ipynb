# loggings

  - it is  a crucial aspect of any application

  - it is used to track the events, errors,  & every information

  - it has built-in logging module offer a flexible frame work  for emtting log messages from python program

```python
import logging

# Configure logging settings
logging.basicConfig(level=logging.DEBUG)

# log meassages of different severity levels
logging.debug("This is a debug message")
logging.info("This is an info message")
logging.warning("This is a warning message")
logging.error("This is an error message")
logging.critical("This is a critical message")
```

- log levels
  1. info

  2. debug 

  3. warning

  4. error

  5. critical

- info:
   - confirmation that things are expected

- example

```python
def add(a,b):
    result = a + b
    logging.info(f"Adding {a} and {b}, result: {result}")
    return result

def subtract(a,b):
    result = a - b
    logging.info(f"Subtracting {b} from {a}, result: {result}")
    return result


def multiply(a,b):
    result = a * b
    logging.info(f"Multiplying {a} and {b}, result: {result}")
    return result    

def divide(a,b):
    if b == 0:
        logging.error("Attempted to divide by zero.")
        raise ValueError("Denominator cannot be zero.")
    result = a / b
    logging.info(f"Dividing {a} by {b}, result: {result}")
    return result 
```

```python
add(10, 5)

subtract(10, 5)

multiply(10, 5)

divide(10, 5)
```
