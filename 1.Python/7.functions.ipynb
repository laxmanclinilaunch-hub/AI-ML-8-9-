```python
def cal_area(l,w):
    """
    Calcuate the area of  Rectangle

    Parameters:
    length (float): the length of the rectangle
    width (float): the width of the rectangle

    returns: 
    float: the area of the rectangle
    """

    return l*w

def cal_perimeter(l,w):
    """
    Calcuate the Perimeter of  Rectangle
    
    Parameters:
        length (float): the length of the rectangle
        width (float): the width of the rectangle
    
        returns: 
        float: the perimeter of the rectangle
    """

    return 2*(l+w)



def print_shape(l,w):
    """
    Calcuate the area & Perimeter of  Rectangle
        
        Parameters:
            length (float): the length of the rectangle
            width (float): the width of the rectangle
        
            returns: 
            float: the area & perimeter of the rectangle
    
    """
    area =cal_area(l,w) 
    perimeter=cal_perimeter(l,w)

    print(f" Rectangle with length {l} and width {w} :")
    print(f"area  {area}")
    print(f"perimeter {perimeter}")
```

```python
print_shape(5,2)
```

```python
print_shape(5,9)
```

```python
print_shape(15,5)
```

- decorators

```python
def deco(func):
    def fun():
        print("************")
        func()
        print("--------------")

    return fun
```

```python
@deco
def hello():
    print("hello clrs students")


hello()    
```

```python
@deco

def hello2():
    print("Hi to all")
```

```python
hello2()
```

```python
def docto_login(func):
    def wrapper():
        print("Doctor verified")
        func()

    return wrapper
```

```python
@docto_login
def view_patient():
    print("patient records are displayed")


view_patient()    
```
