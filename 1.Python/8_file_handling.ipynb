- writing of the files

```python
# if the file was not presented

f=open("files/sample.txt","w")

f.write("Hello arunachalm")

f.close()
```

```python
# write the multiple lines


f=open("files/sample.txt",'w')

f.write("Hello pratik")

f.write("\n How are you?")

f.close()
```

```python
# introducing append

f=open('files/sample.txt',"a")
f.write("\n I am fine")
f.close()
```

```python
# write lines 

L= [ "hello\n","Hi\n","How are you\n",'I am fine' ]

f=open("files/sample1.txt",'w')
f.writelines(L)
f.close()
```

- Reading from files

```python
f=open("files/sample.txt","r")

s=f.read()
print(s)
f.close()
```

```python
#reading upto n chars

f=open('files/sample1.txt',"r")

s=f.read(20)
print(s)

f.close()
```
