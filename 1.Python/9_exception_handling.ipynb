      - Example :
      
         - Divide by 0 

         - Database error

```python
def divide(a, b):
    return a / b
```

```python
divide(10, 0)
```

- we can handle this type of error try , except 

```python
def divide(a, b):
    if b == 0:
        raise Exception("we Cannot divide any number with zero")
    else:
        return a / b
```

```python
try:
    result = divide(10, 0)
    print(result)
except Exception as e:
    print("An error occurred:", e)    
```

```python
def divide(a, b):
    try:
        return a / b
    except ZeroDivisionError:
        print("Error: Division by zero is not allowed.")
        return None
```

```python
divide(10, 0)
```

```python
divide(10, 2)
```

```python
def divi(a,b):
    try:
        return a/b
    except Exception as e:
        print("An error occurred:", e)
```

```python
divi(10,5)
```

```python
divi(10,0)
```
