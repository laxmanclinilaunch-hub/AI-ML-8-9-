```python
# split of data


from statsmodels.stats.outliers_influence import variance_inflation_factor

def calc_vif(X):

    # Calculating VIF
    vif = pd.DataFrame()
    vif["variables"] = X.columns
    vif["VIF"] = [variance_inflation_factor(X.values, i) for i in range(X.shape[1])]

    return(vif)
```

# Feature scaling

```python
from sklearn.model_selection import train_test_split
```

```python
X_train,X_test, y_train,y_test=train_test_split(X,y,test_size=0.25,random_state=117,stratify=y)
```

```python
X_train.shape
```

```python
X_test.shape
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()


X_train_scaled = scaler.fit_transform(X_train)
```

```python
X_test_scaled = scaler.transform(X_test)
```

# model traning

```python
# 

from sklearn.linear_model import LogisticRegression

lr= LogisticRegression()

lr.fit(X_train_scaled, y_train)     # training the model
```

```python
y_pred=lr.predict(X_test_scaled)


# probability of claSS-1


y_prob=lr.predict_proba(X_test_scaled)[:,1]
```

```python
y_pred     # our prediction  done by machine 
```

```python
y_test    # our actial values
```

```python
result =pd.DataFrame(
    {"actual":y_test,
     "predicted":y_pred,
     "y_cap":y_prob}
)

result
```
