- class method:

   - which can be defined with using the decorator @classmethod and along with you need to use cls parameter and using class variables 

```python
class netflix:
    new_show="kalki"


    def __init__(self,mail,password):
        self.mail=mail
        self.password=password
        self.service="Free"
        self.amount=0


    def get_password(self):
        return self.password


    @classmethod
    def get_newshow(cls):
        return cls.new_show


    @classmethod
    def change_newshow(cls,show_name):
        cls.new_show=show_name
```

```python
netflix.get_newshow()    # accessing of class method using classname 
```

```python
user=netflix("lakshman@gmail.com",1234)    # accesing class method using object

```

```python
user.get_newshow()
```

```python
netflix.change_newshow("salar")
                                    # modifying of classvariable using  class nane 
print(netflix.get_newshow())
```

```python
user=netflix("lakshman@gmail.com",1234) 


user.change_newshow("kalki-2")                      # modifying of classvariable using object


user.get_newshow()
```

- Static method :

          - static method can be called without self or cls

          - this method can access any attribute like either class attribute or instance attribute 

          - this can be written decorator @static method



```python
class mymath:
    @staticmethod
    def add(x,y):
        return x+y
```

```python
mymath.add(3,5)
```

```python
class netflex:

    new_show="kalki"


    def __init__(self,mail,password):
        self.mail=mail
        self.password=password
        self.service="free"
        self.amount=0



    def get_password(self):
        return self.password


    @staticmethod
    def get_newshow():
        return netflex.new_show

    
    @staticmethod
    def get_mail(instance):
        return instance.mail

         
        
```

# - encapsulation:

- it is  a pillar of oops
- it using varibles and methods we have to controlling access to the data

-  a core concept in programming where data and the functions that use that data are bundled together into a single unit





- public memeber:

  - it can access anywhere (inside the class or outside of the class)

           - self.name=name



- proteced member:


-  the protected memeber start with one underscore
 - it can access inside the class and its subclass ( child class)

        - self._name = name 

- private memeber
     - private memeber  start with duble underscore

     - the private memeber can't access directly 

     - to access private member we have to (getter)  method and for modification use (setter) method 

                   - self.__name = name 





- public member

```python
class mentor:
    def __init__(self):
        self.name="laxman"
        


obj=mentor()

print(obj.name)
```

- protected member

```python
class mentor:
    def __init__(self):
        self._name='laxman'
        

obj=mentor()

print(obj._name)  # protected member

```

- protected member with subclasses

```python
class father:
    def __init__(self):
        self._salary=70000




class son(father):
    def display(self):
        print(self._salary)



obj=son()


obj.display()
```

- private member

```python
class mentor:
    def __init__(self):
        self.__name="laxman"



obj=mentor()

print(obj.__name)
```

- for accessing of private member we have to use getter method

```python
class mentor:
    def __init__(self):
        self.__name="laxman"


    def get_mentorname(self):
            return self.__name


obj=mentor()

print(obj.get_mentorname())
```

- For modifying private data using setter method

```python
class mentor:
    def __init__(self):
            self.__name="laxman"

    def set_mentorname(self,change):
          self.__name=change


    def get_mentorname(self):
          return  self.__name   


obj=mentor()


obj.set_mentorname("arunachalam")

print(obj.get_mentorname())
```

example

```python
class patient:
    def __init__(self):
        self.name="arunachalam"   # public member 
        self._blood_group="A1 NEGATIVE"   # protected member 
        self.__medical_report="CANCER"     # private

    def get_medical_report(self):
        return self.__medical_report


obj=patient()


print(obj.name)   # accessing of public member
print(obj._blood_group)   # accessing of protected member
print(obj.get_medical_report())  # accessing of private member
```
