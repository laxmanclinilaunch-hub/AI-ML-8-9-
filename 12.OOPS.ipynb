```python
class pratik:
    sub="AI&ML"      # class attribute 

    def __init__(self,name, age):     # constructor
        # instance  attributes   
        self.name =name
        self.age=age


    def find_name(self):       # method (function)
        return self.name

    def find_age(self):        # method
        return self.age   
```

- Type of constructor

- Non-parameterized constructor:

        - initilizes attributes with default values 

```python
class hotstar:
    new_show="kalki"

    def __init__(self):
        self.mail="laxman@gmail.com"
        self.password=5630                          # Non-parameterized constructor
        self.service="Free"
        self.amount=0

    def get_password(self):
        return self.password     
```

- parameterized constructor:
  - it takes one or more arguments  in addition to self 

```python
class hotstar:
    new_show="kalki"

    def __init__(self,mail,password,service,amount):
        self.mail=mail
        self.password=password                      # parameterized constructor
        self.service=service
        self.amount=amount

    def get_password(self):
        return self.password  
```
